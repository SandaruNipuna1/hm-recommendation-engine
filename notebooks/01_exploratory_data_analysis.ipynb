```python
from tqdm import tqdm

transaction_file = DATA_DIR / "transactions_train.csv"

total_rows = 0
earliest_date = None
latest_date = None

missing_counts = {
    "t_dat": 0,
    "customer_id": 0,
    "article_id": 0,
    "price": 0,
    "sales_channel_id": 0
}

for chunk in tqdm(
    pd.read_csv(transaction_file, chunksize=500_000),
    desc="Processing transactions"
):
    total_rows += len(chunk)

    chunk_min = chunk["t_dat"].min()
    chunk_max = chunk["t_dat"].max()

    earliest_date = min(earliest_date, chunk_min) if earliest_date else chunk_min
    latest_date = max(latest_date, chunk_max) if latest_date else chunk_max

    for column in missing_counts:
        missing_counts[column] += chunk[column].isna().sum()

print("\nTRANSACTION DATASET SUMMARY")
print("Total transactions:", f"{total_rows:,}")
print("Earliest transaction:", earliest_date)
print("Latest transaction:", latest_date)

print("\nMISSING VALUES")
display(pd.Series(missing_counts, name="missing_count").to_frame())
```

```python
from collections import Counter

daily_transactions = Counter()
channel_counts = Counter()

price_sum = 0
price_min = float("inf")
price_max = float("-inf")
price_count = 0

for chunk in tqdm(
    pd.read_csv(transaction_file, chunksize=500_000),
    desc="Analysing transaction trends"
):
    daily_transactions.update(
        chunk["t_dat"].value_counts().to_dict()
    )

    channel_counts.update(
        chunk["sales_channel_id"].value_counts().to_dict()
    )

    price_sum += chunk["price"].sum()
    price_min = min(price_min, chunk["price"].min())
    price_max = max(price_max, chunk["price"].max())
    price_count += chunk["price"].count()

daily_counts = pd.Series(daily_transactions).sort_index()

print("DAILY TRANSACTION STATISTICS")
display(daily_counts.describe())

print("\nSALES CHANNEL DISTRIBUTION")
display(
    pd.Series(channel_counts, name="transactions")
    .sort_index()
)

print("\nPRICE STATISTICS")
print("Mean price:", price_sum / price_count)
print("Minimum price:", price_min)
print("Maximum price:", price_max)
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Convert daily transaction counts into a time series
daily_counts.index = pd.to_datetime(daily_counts.index)
daily_counts = daily_counts.sort_index()

# Calculate 7-day rolling average
rolling_7d = daily_counts.rolling(window=7).mean()

# Aggregate transactions by month
monthly_counts = daily_counts.resample("MS").sum()

# Daily transaction trends
plt.figure(figsize=(16, 6))

plt.plot(
    daily_counts.index,
    daily_counts.values,
    alpha=0.35,
    label="Daily transactions"
)

plt.plot(
    rolling_7d.index,
    rolling_7d.values,
    linewidth=2,
    label="7-day rolling average"
)

plt.title("H&M Daily Transaction Trends (2018–2020)")
plt.xlabel("Date")
plt.ylabel("Number of transactions")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()

# Monthly transaction trends
plt.figure(figsize=(14, 5))

plt.plot(
    monthly_counts.index,
    monthly_counts.values,
    marker="o"
)

plt.title("H&M Monthly Transaction Volume (2018–2020)")
plt.xlabel("Month")
plt.ylabel("Number of transactions")
plt.grid(alpha=0.25)
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%b %Y"))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
monthly_summary = monthly_counts.to_frame(
    name="transactions"
)

monthly_summary["year"] = monthly_summary.index.year
monthly_summary["month"] = monthly_summary.index.month

print("TOP 10 MONTHS BY TRANSACTION VOLUME")

display(
    monthly_summary.nlargest(10, "transactions")
)

print("\nBOTTOM 10 MONTHS BY TRANSACTION VOLUME")

display(
    monthly_summary.nsmallest(10, "transactions")
)
```

```python
# Create a calendar-month comparison
monthly_comparison = monthly_summary.pivot(
    index="month",
    columns="year",
    values="transactions"
)

# Exclude September because both observations are partial months
monthly_comparison = monthly_comparison.drop(index=9)

display(monthly_comparison)

# Visualize month-to-month comparison
monthly_comparison.plot(
    kind="bar",
    figsize=(13, 6),
    width=0.8
)

plt.title("Monthly Transaction Comparison Across Years")
plt.xlabel("Calendar Month")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.legend(title="Year")
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
# Compare 2019 and 2020 for January–August
yoy_comparison = monthly_comparison[[2019, 2020]].dropna()

yoy_comparison["change"] = (
    yoy_comparison[2020] - yoy_comparison[2019]
)

yoy_comparison["change_percentage"] = (
    yoy_comparison["change"] / yoy_comparison[2019]
) * 100

display(yoy_comparison.round(2))

print("\nAVERAGE YEAR-OVER-YEAR CHANGE:")
print(
    round(yoy_comparison["change_percentage"].mean(), 2),
    "%"
)
```

```python
from collections import Counter

customer_frequency = Counter()

for chunk in tqdm(
    pd.read_csv(
        transaction_file,
        chunksize=500_000,
        usecols=["customer_id"]
    ),
    desc="Analysing customer activity"
):
    customer_frequency.update(
        chunk["customer_id"].value_counts().to_dict()
    )

customer_activity = pd.Series(
    customer_frequency,
    name="transaction_lines"
)

print("CUSTOMER ACTIVITY SUMMARY")
print("Unique purchasing customers:", f"{len(customer_activity):,}")

display(customer_activity.describe())

print("\nTOP 10 MOST ACTIVE CUSTOMERS")
display(customer_activity.nlargest(10))

print("\nBOTTOM 10 LEAST ACTIVE CUSTOMERS")
display(customer_activity.nsmallest(10))
```

```python
# Group customers by transaction-line frequency

activity_groups = pd.cut(
    customer_activity,
    bins=[0, 1, 5, 10, 20, 50, float("inf")],
    labels=[
        "1 transaction",
        "2–5 transactions",
        "6–10 transactions",
        "11–20 transactions",
        "21–50 transactions",
        "More than 50"
    ]
)

activity_summary = (
    activity_groups
    .value_counts(sort=False)
    .to_frame(name="customers")
)

activity_summary["percentage"] = (
    activity_summary["customers"]
    / len(customer_activity) * 100
)

display(activity_summary.round(2))

# Visualize customer activity distribution
activity_summary["customers"].plot(
    kind="bar",
    figsize=(11, 5)
)

plt.title("Customer Distribution by Transaction Activity")
plt.xlabel("Transaction-line frequency")
plt.ylabel("Number of customers")
plt.xticks(rotation=30)
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
# Calculate transaction-line contribution by customer segment

contribution = customer_activity.groupby(activity_groups).sum()

contribution_summary = pd.DataFrame({
    "customers": activity_groups.value_counts(sort=False),
    "transaction_lines": contribution
})

contribution_summary["customer_percentage"] = (
    contribution_summary["customers"]
    / len(customer_activity) * 100
)

contribution_summary["transaction_percentage"] = (
    contribution_summary["transaction_lines"]
    / customer_activity.sum() * 100
)

display(contribution_summary.round(2))
```

```python
from collections import Counter

article_frequency = Counter()

for chunk in tqdm(
    pd.read_csv(
        transaction_file,
        chunksize=500_000,
        usecols=["article_id"]
    ),
    desc="Analysing article popularity"
):
    article_frequency.update(
        chunk["article_id"].value_counts().to_dict()
    )

article_activity = pd.Series(
    article_frequency,
    name="transaction_lines"
)

print("PRODUCT POPULARITY SUMMARY")
print("Unique purchased articles:", f"{len(article_activity):,}")

display(article_activity.describe())

print("\nTOP 20 MOST PURCHASED ARTICLES")
display(article_activity.nlargest(20))
```

```python
import matplotlib.pyplot as plt
import numpy as np

sorted_popularity = article_activity.sort_values(
    ascending=False
).reset_index(drop=True)

plt.figure(figsize=(12, 6))

plt.plot(
    np.arange(1, len(sorted_popularity) + 1),
    sorted_popularity.values
)

plt.yscale("log")

plt.title("H&M Article Popularity Distribution")
plt.xlabel("Articles (ranked by popularity)")
plt.ylabel("Number of transaction lines (log scale)")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Calculate popularity concentration

total_transactions = article_activity.sum()
total_articles = len(article_activity)

percentages = [1, 5, 10, 20]

print("ARTICLE POPULARITY CONCENTRATION")
print("-" * 55)

for percentage in percentages:

    number_of_articles = int(
        np.ceil(total_articles * percentage / 100)
    )

    top_transactions = (
        sorted_popularity.iloc[:number_of_articles].sum()
    )

    transaction_share = (
        top_transactions / total_transactions * 100
    )

    print(
        f"Top {percentage}% of articles "
        f"({number_of_articles:,} articles): "
        f"{transaction_share:.2f}% of transactions"
    )
```

```python
# Connect article popularity with product metadata

article_popularity = (
    article_activity
    .rename("transaction_lines")
    .reset_index()
)

article_popularity.columns = [
    "article_id",
    "transaction_lines"
]

article_details = article_popularity.merge(
    articles[[
        "article_id",
        "product_group_name",
        "product_type_name"
    ]],
    on="article_id",
    how="left"
)

# Calculate popularity by product group

group_popularity = (
    article_details
    .groupby("product_group_name")["transaction_lines"]
    .sum()
    .sort_values(ascending=False)
)

group_share = (
    group_popularity / group_popularity.sum() * 100
)

group_summary = pd.DataFrame({
    "transaction_lines": group_popularity,
    "percentage": group_share
})

display(group_summary.round(2))
```

```python
# Analyse popularity by product type

type_popularity = (
    article_details
    .groupby("product_type_name")["transaction_lines"]
    .sum()
    .sort_values(ascending=False)
)

top_20_types = type_popularity.head(20)

type_summary = pd.DataFrame({
    "transaction_lines": top_20_types,
    "percentage": (
        top_20_types / article_activity.sum() * 100
    )
})

display(type_summary.round(2))

# Visualize top 20 product types

plt.figure(figsize=(12, 8))

top_20_types.sort_values().plot(kind="barh")

plt.title("Top 20 Most Purchased H&M Product Types")
plt.xlabel("Number of transaction lines")
plt.ylabel("Product type")

plt.tight_layout()
plt.show()
```

```python
# Combine customer ages with purchasing activity

customer_purchases = (
    customer_activity
    .rename("transaction_lines")
    .reset_index()
)

customer_purchases.columns = [
    "customer_id",
    "transaction_lines"
]

customer_age_activity = customers[
    ["customer_id", "age"]
].merge(
    customer_purchases,
    on="customer_id",
    how="inner"
)

# Create age groups

age_bins = [0, 20, 30, 40, 50, 60, 100]

age_labels = [
    "Under 20",
    "20-29",
    "30-39",
    "40-49",
    "50-59",
    "60+"
]

customer_age_activity["age_group"] = pd.cut(
    customer_age_activity["age"],
    bins=age_bins,
    labels=age_labels,
    right=False
)

customer_age_activity["age_group"] = (
    customer_age_activity["age_group"]
    .cat.add_categories(["Unknown"])
    .fillna("Unknown")
)

# Aggregate purchasing behaviour

age_summary = (
    customer_age_activity
    .groupby("age_group", observed=False)
    .agg(
        purchasing_customers=("customer_id", "nunique"),
        transaction_lines=("transaction_lines", "sum"),
        avg_transactions_per_customer=(
            "transaction_lines", "mean"
        )
    )
)

age_summary["transaction_percentage"] = (
    age_summary["transaction_lines"]
    / age_summary["transaction_lines"].sum() * 100
)

display(age_summary.round(2))
```

```python
# Merge transactions with article product categories

age_product_data = transactions_sample.merge(
    articles[["article_id", "product_group_name"]],
    on="article_id",
    how="left"
)

# Attach customer age groups

age_product_data = age_product_data.merge(
    customer_age_activity[["customer_id", "age_group"]],
    on="customer_id",
    how="left"
)

# Calculate product category distribution by age group

age_category_summary = (
    age_product_data
    .groupby(
        ["age_group", "product_group_name"],
        observed=True
    )
    .size()
    .reset_index(name="transaction_lines")
)

age_category_summary["percentage"] = (
    age_category_summary["transaction_lines"]
    / age_category_summary.groupby(
        "age_group"
    )["transaction_lines"].transform("sum")
    * 100
)

# Display top 5 categories for each age group

top_categories = (
    age_category_summary
    .sort_values(
        ["age_group", "transaction_lines"],
        ascending=[True, False]
    )
    .groupby("age_group", observed=True)
    .head(5)
)

display(top_categories.round(2))
```
