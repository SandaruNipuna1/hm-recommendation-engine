```python
from tqdm import tqdm

transaction_file = DATA_DIR / "transactions_train.csv"

total_rows = 0
earliest_date = None
latest_date = None

missing_counts = {
    "t_dat": 0,
    "customer_id": 0,
    "article_id": 0,
    "price": 0,
    "sales_channel_id": 0
}

for chunk in tqdm(
    pd.read_csv(transaction_file, chunksize=500_000),
    desc="Processing transactions"
):
    total_rows += len(chunk)

    chunk_min = chunk["t_dat"].min()
    chunk_max = chunk["t_dat"].max()

    earliest_date = min(earliest_date, chunk_min) if earliest_date else chunk_min
    latest_date = max(latest_date, chunk_max) if latest_date else chunk_max

    for column in missing_counts:
        missing_counts[column] += chunk[column].isna().sum()

print("\nTRANSACTION DATASET SUMMARY")
print("Total transactions:", f"{total_rows:,}")
print("Earliest transaction:", earliest_date)
print("Latest transaction:", latest_date)

print("\nMISSING VALUES")
display(pd.Series(missing_counts, name="missing_count").to_frame())
```

```python
from collections import Counter

daily_transactions = Counter()
channel_counts = Counter()

price_sum = 0
price_min = float("inf")
price_max = float("-inf")
price_count = 0

for chunk in tqdm(
    pd.read_csv(transaction_file, chunksize=500_000),
    desc="Analysing transaction trends"
):
    daily_transactions.update(
        chunk["t_dat"].value_counts().to_dict()
    )

    channel_counts.update(
        chunk["sales_channel_id"].value_counts().to_dict()
    )

    price_sum += chunk["price"].sum()
    price_min = min(price_min, chunk["price"].min())
    price_max = max(price_max, chunk["price"].max())
    price_count += chunk["price"].count()

daily_counts = pd.Series(daily_transactions).sort_index()

print("DAILY TRANSACTION STATISTICS")
display(daily_counts.describe())

print("\nSALES CHANNEL DISTRIBUTION")
display(
    pd.Series(channel_counts, name="transactions")
    .sort_index()
)

print("\nPRICE STATISTICS")
print("Mean price:", price_sum / price_count)
print("Minimum price:", price_min)
print("Maximum price:", price_max)
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Convert daily transaction counts into a time series
daily_counts.index = pd.to_datetime(daily_counts.index)
daily_counts = daily_counts.sort_index()

# Calculate 7-day rolling average
rolling_7d = daily_counts.rolling(window=7).mean()

# Aggregate transactions by month
monthly_counts = daily_counts.resample("MS").sum()

# Daily transaction trends
plt.figure(figsize=(16, 6))

plt.plot(
    daily_counts.index,
    daily_counts.values,
    alpha=0.35,
    label="Daily transactions"
)

plt.plot(
    rolling_7d.index,
    rolling_7d.values,
    linewidth=2,
    label="7-day rolling average"
)

plt.title("H&M Daily Transaction Trends (2018–2020)")
plt.xlabel("Date")
plt.ylabel("Number of transactions")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()

# Monthly transaction trends
plt.figure(figsize=(14, 5))

plt.plot(
    monthly_counts.index,
    monthly_counts.values,
    marker="o"
)

plt.title("H&M Monthly Transaction Volume (2018–2020)")
plt.xlabel("Month")
plt.ylabel("Number of transactions")
plt.grid(alpha=0.25)
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%b %Y"))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
from collections import Counter

customer_frequency = Counter()

for chunk in tqdm(
    pd.read_csv(
        transaction_file,
        chunksize=500_000,
        usecols=["customer_id"]
    ),
    desc="Analysing customer activity"
):
    customer_frequency.update(
        chunk["customer_id"].value_counts().to_dict()
    )

customer_activity = pd.Series(
    customer_frequency,
    name="transaction_lines"
)

print("CUSTOMER ACTIVITY SUMMARY")
print("Unique purchasing customers:", f"{len(customer_activity):,}")

display(customer_activity.describe())

print("\nTOP 10 MOST ACTIVE CUSTOMERS")
display(customer_activity.nlargest(10))

print("\nBOTTOM 10 LEAST ACTIVE CUSTOMERS")
display(customer_activity.nsmallest(10))
```

```python
# Group customers by transaction-line frequency

activity_groups = pd.cut(
    customer_activity,
    bins=[0, 1, 5, 10, 20, 50, float("inf")],
    labels=[
        "1 transaction",
        "2–5 transactions",
        "6–10 transactions",
        "11–20 transactions",
        "21–50 transactions",
        "More than 50"
    ]
)

activity_summary = (
    activity_groups
    .value_counts(sort=False)
    .to_frame(name="customers")
)

activity_summary["percentage"] = (
    activity_summary["customers"]
    / len(customer_activity) * 100
)

display(activity_summary.round(2))

# Visualize customer activity distribution
activity_summary["customers"].plot(
    kind="bar",
    figsize=(11, 5)
)

plt.title("Customer Distribution by Transaction Activity")
plt.xlabel("Transaction-line frequency")
plt.ylabel("Number of customers")
plt.xticks(rotation=30)
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
# Calculate transaction-line contribution by customer segment

contribution = customer_activity.groupby(activity_groups).sum()

contribution_summary = pd.DataFrame({
    "customers": activity_groups.value_counts(sort=False),
    "transaction_lines": contribution
})

contribution_summary["customer_percentage"] = (
    contribution_summary["customers"]
    / len(customer_activity) * 100
)

contribution_summary["transaction_percentage"] = (
    contribution_summary["transaction_lines"]
    / customer_activity.sum() * 100
)

display(contribution_summary.round(2))
```

```python
# Combine customer ages with purchasing activity

customer_purchases = (
    customer_activity
    .rename("transaction_lines")
    .reset_index()
)

customer_purchases.columns = [
    "customer_id",
    "transaction_lines"
]

customer_age_activity = customers[
    ["customer_id", "age"]
].merge(
    customer_purchases,
    on="customer_id",
    how="inner"
)

# Create age groups

age_bins = [0, 20, 30, 40, 50, 60, 100]

age_labels = [
    "Under 20",
    "20-29",
    "30-39",
    "40-49",
    "50-59",
    "60+"
]

customer_age_activity["age_group"] = pd.cut(
    customer_age_activity["age"],
    bins=age_bins,
    labels=age_labels,
    right=False
)

customer_age_activity["age_group"] = (
    customer_age_activity["age_group"]
    .cat.add_categories(["Unknown"])
    .fillna("Unknown")
)

# Aggregate purchasing behaviour

age_summary = (
    customer_age_activity
    .groupby("age_group", observed=False)
    .agg(
        purchasing_customers=("customer_id", "nunique"),
        transaction_lines=("transaction_lines", "sum"),
        avg_transactions_per_customer=(
            "transaction_lines", "mean"
        )
    )
)

age_summary["transaction_percentage"] = (
    age_summary["transaction_lines"]
    / age_summary["transaction_lines"].sum() * 100
)

display(age_summary.round(2))
```

```python
import pandas as pd

transactions = pd.read_csv(
    "../data/raw/transactions_train.csv",
    usecols=["t_dat", "customer_id", "article_id"],
    dtype={
        "customer_id": "string",
        "article_id": "int32"
    },
    parse_dates=["t_dat"]
)

print("Original dataset shape:", transactions.shape)
print("Date range:", transactions["t_dat"].min(), "to", transactions["t_dat"].max())
print("Date data type:", transactions["t_dat"].dtype)
```

```python
monthly_transactions = (
    transactions
    .groupby(transactions["t_dat"].dt.to_period("M"))
    .size()
)

display(monthly_transactions.to_frame("transaction_count"))

monthly_transactions.plot(
    figsize=(14, 5),
    title="Monthly Transaction Volume"
)

plt.xlabel("Month")
plt.ylabel("Number of Transactions")
plt.grid(alpha=0.3)
plt.show()
```

```python
train = transactions[
    transactions["t_dat"] <= "2020-07-31"
]

validation = transactions[
    (transactions["t_dat"] >= "2020-08-01") &
    (transactions["t_dat"] <= "2020-08-31")
]

test = transactions[
    transactions["t_dat"] >= "2020-09-01"
]

print("TRAIN")
print("Transactions:", len(train))
print("Customers:", train["customer_id"].nunique())

print("\nVALIDATION")
print("Transactions:", len(validation))
print("Customers:", validation["customer_id"].nunique())

print("\nTEST")
print("Transactions:", len(test))
print("Customers:", test["customer_id"].nunique())
```

```python
train_customers = set(train["customer_id"].dropna().unique())
validation_customers = set(validation["customer_id"].dropna().unique())
test_customers = set(test["customer_id"].dropna().unique())

# Customers appearing in multiple periods
validation_existing = validation_customers & train_customers
test_existing = test_customers & train_customers

# New customers not seen during training
validation_new = validation_customers - train_customers
test_new = test_customers - train_customers

print("CUSTOMER OVERLAP ANALYSIS")
print("-" * 40)

print("Validation customers:", len(validation_customers))
print("Previously seen:", len(validation_existing))
print("New customers:", len(validation_new))

print("\nTest customers:", len(test_customers))
print("Previously seen:", len(test_existing))
print("New customers:", len(test_new))

print("\nPercentage of test customers seen during training:",
      round(len(test_existing) / len(test_customers) * 100, 2), "%")
```

```python
# Customers known after training + validation
train_validation_customers = (
    train_customers | validation_customers
)

test_known_after_validation = (
    test_customers & train_validation_customers
)

test_new_after_validation = (
    test_customers - train_validation_customers
)

print("FINAL TEST CUSTOMER OVERLAP")
print("-" * 40)

print("Total test customers:", len(test_customers))

print(
    "Known from training + validation:",
    len(test_known_after_validation)
)

print(
    "New even after validation:",
    len(test_new_after_validation)
)

print(
    "Known customer percentage:",
    round(
        len(test_known_after_validation) /
        len(test_customers) * 100, 2
    ),
    "%"
)
```
